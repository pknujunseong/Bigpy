```python
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf

from tensorflow.keras.datasets import fashion_mnist
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import (
    Input, Flatten, Dense, Dropout, BatchNormalization, Activation
)
from tensorflow.keras.callbacks import EarlyStopping
from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    accuracy_score, classification_report, ConfusionMatrixDisplay
)

# 1. Random seed
tf.keras.utils.set_random_seed(123)

# 2. Load dataset
(X_train_full, y_train_full), (X_test, y_test) = fashion_mnist.load_data()

# 3. Data preprocessing
X_train_full = X_train_full.astype('float32') / 255.0
X_test = X_test.astype('float32') / 255.0

X_train, X_val, y_train, y_val = train_test_split(
    X_train_full, y_train_full,
    test_size=0.2,
    random_state=123,
    stratify=y_train_full
)

# 4. DNN model
model = Sequential([
    Input(shape=(28, 28)),
    Flatten(),

    Dense(256, use_bias=False),
    BatchNormalization(),
    Activation('relu'),

    Dense(128, use_bias=False),
    BatchNormalization(),
    Activation('relu'),

    Dropout(0.25),
    Dense(10, activation='softmax')
])

model.summary()

# 5. Model compile
model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

# 6. Model training
history = model.fit(
    X_train, y_train,
    validation_data=(X_val, y_val),
    epochs=100,
    batch_size=128,
    callbacks=[
        EarlyStopping(
            monitor='val_loss',
            patience=15,
            restore_best_weights=True
        )
    ]
)

# 7. Prediction
y_pred = np.argmax(model.predict(X_test), axis=1)


# 8. Classification results
test_loss, test_accuracy = model.evaluate(X_test, y_test)

print("Test Loss:", test_loss)
print("Test Accuracy:", test_accuracy)
print(classification_report(y_test, y_pred))

# 9. Confusion matrix
ConfusionMatrixDisplay.from_predictions(y_test, y_pred, cmap='Blues')
plt.title('DNN Confusion Matrix')
plt.show()

# 10. Loss curve
plt.plot(history.history['loss'], label='Training Loss')
plt.plot(history.history['val_loss'], label='Validation Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()
plt.show()

# 11. Optimal epoch
print("Optimal Epoch:", np.argmin(history.history['val_loss']) + 1)
```

```python
y_pred_dnn = np.argmax(model.predict(X_test), axis=1)
```

```python
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf

from tensorflow.keras.datasets import fashion_mnist
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import (
    Input, Conv2D, MaxPooling2D, Flatten, Dense,
    Dropout, BatchNormalization, Activation
)
from tensorflow.keras.callbacks import EarlyStopping
from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    accuracy_score, classification_report, ConfusionMatrixDisplay
)

# 1. Random seed
tf.keras.utils.set_random_seed(123)

# 2. Load dataset
(X_train_full, y_train_full), (X_test, y_test) = fashion_mnist.load_data()

# 3. Data preprocessing
X_train_full = X_train_full.astype('float32') / 255.0
X_test = X_test.astype('float32') / 255.0

X_train_full = X_train_full.reshape(-1, 28, 28, 1)
X_test = X_test.reshape(-1, 28, 28, 1)

X_train, X_val, y_train, y_val = train_test_split(
    X_train_full, y_train_full,
    test_size=0.2,
    random_state=123,
    stratify=y_train_full
)

# 4. CNN model
model = Sequential([
    Input(shape=(28, 28, 1)),

    Conv2D(32, kernel_size=(3, 3), use_bias=False),
    BatchNormalization(),
    Activation('relu'),
    MaxPooling2D(pool_size=(2, 2)),

    Conv2D(64, kernel_size=(3, 3), use_bias=False),
    BatchNormalization(),
    Activation('relu'),
    MaxPooling2D(pool_size=(2, 2)),

    Flatten(),

    Dense(128, use_bias=False),
    BatchNormalization(),
    Activation('relu'),
    Dropout(0.25),

    Dense(10, activation='softmax')
])

model.summary()

# 5. Model compile
model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

# 6. Model training
history = model.fit(
    X_train, y_train,
    validation_data=(X_val, y_val),
    epochs=100,
    batch_size=128,
    callbacks=[
        EarlyStopping(
            monitor='val_loss',
            patience=15,
            restore_best_weights=True
        )
    ]
)

# 7. Prediction
y_pred = np.argmax(model.predict(X_test), axis=1)

# 8. Classification results
test_loss, test_accuracy = model.evaluate(X_test, y_test)

print("Test Loss:", test_loss)
print("Test Accuracy:", test_accuracy)
print(classification_report(y_test, y_pred))

# 9. Confusion matrix
ConfusionMatrixDisplay.from_predictions(y_test, y_pred, cmap='Blues')
plt.title('CNN Confusion Matrix')
plt.show()

# 10. Loss curve
plt.plot(history.history['loss'], label='Training Loss')
plt.plot(history.history['val_loss'], label='Validation Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()
plt.show()

# 11. Optimal epoch
print("Optimal Epoch:", np.argmin(history.history['val_loss']) + 1)
```

#2. CIFAR-10
DNN과 CNN 모두 Batch Normalization 및 ResNet을 적용합니다.

```python
, import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf

from tensorflow.keras.datasets import cifar10
from tensorflow.keras import layers, models
from tensorflow.keras.callbacks import EarlyStopping
from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    classification_report, ConfusionMatrixDisplay
)

# 1. Random seed
tf.keras.utils.set_random_seed(123)

# 2. Load dataset
(X_train_full, y_train_full), (X_test, y_test) = cifar10.load_data()

# 3. Data preprocessing
X_train_full = X_train_full.astype('float32') / 255.0
X_test = X_test.astype('float32') / 255.0

y_train_full = y_train_full.ravel()
y_test = y_test.ravel()

X_train, X_val, y_train, y_val = train_test_split(
    X_train_full, y_train_full,
    test_size=0.2,
    random_state=123,
    stratify=y_train_full
)

# 4. Residual block
def residual_fc_block(x, units):

    shortcut = x

    x = layers.Dense(units, use_bias=False)(x)
    x = layers.BatchNormalization()(x)
    x = layers.Activation('relu')(x)

    x = layers.Dense(units, use_bias=False)(x)
    x = layers.BatchNormalization()(x)

    x = layers.Add()([x, shortcut])
    x = layers.Activation('relu')(x)

    return x

# 5. DNN construction
inputs = layers.Input(shape=(32, 32, 3))

x = layers.Flatten()(inputs)

x = layers.Dense(256, use_bias=False)(x)
x = layers.BatchNormalization()(x)
x = layers.Activation('relu')(x)

x = residual_fc_block(x, 256)
x = residual_fc_block(x, 256)

x = layers.Dropout(0.25)(x)

outputs = layers.Dense(10, activation='softmax')(x)

model_dnn = models.Model(inputs, outputs)

model_dnn.summary()

# 6. Compile
model_dnn.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

# 7. Training
history_dnn = model_dnn.fit(
    X_train, y_train,
    validation_data=(X_val, y_val),
    epochs=100,
    batch_size=128,
    callbacks=[
        EarlyStopping(
            monitor='val_loss',
            patience=15,
            restore_best_weights=True
        )
    ]
)

# 8. Prediction
y_pred_dnn = np.argmax(model_dnn.predict(X_test), axis=1)

# 9. Classification results
loss_dnn, acc_dnn = model_dnn.evaluate(X_test, y_test)

print("Test Loss:", loss_dnn)
print("Test Accuracy:", acc_dnn)
print(classification_report(y_test, y_pred_dnn))

# 10. Confusion matrix
ConfusionMatrixDisplay.from_predictions(
    y_test, y_pred_dnn, cmap='Blues'
)
plt.title('CIFAR-10 DNN Confusion Matrix')
plt.show()

# 11. Loss curve
plt.plot(history_dnn.history['loss'], label='Training Loss')
plt.plot(history_dnn.history['val_loss'], label='Validation Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()
plt.show()

print("Optimal Epoch:",
      np.argmin(history_dnn.history['val_loss']) + 1)
```

2-2) CNN + Batch Normalization + ResNet

```python
# 1. Random seed
tf.keras.utils.set_random_seed(123)

# 2. Residual block
def residual_conv_block(x, filters):

    shortcut = x

    x = layers.Conv2D(
        filters, 3,
        padding='same',
        use_bias=False
    )(x)

    x = layers.BatchNormalization()(x)
    x = layers.Activation('relu')(x)

    x = layers.Conv2D(
        filters, 3,
        padding='same',
        use_bias=False
    )(x)

    x = layers.BatchNormalization()(x)

    x = layers.Add()([x, shortcut])
    x = layers.Activation('relu')(x)

    return x

# 3. CNN construction
inputs = layers.Input(shape=(32, 32, 3))

x = layers.Conv2D(
    32, 3, padding='same', use_bias=False
)(inputs)

x = layers.BatchNormalization()(x)
x = layers.Activation('relu')(x)

x = residual_conv_block(x, 32)
x = layers.MaxPooling2D(2)(x)

x = layers.Conv2D(
    64, 3, padding='same', use_bias=False
)(x)

x = layers.BatchNormalization()(x)
x = layers.Activation('relu')(x)

x = residual_conv_block(x, 64)
x = layers.MaxPooling2D(2)(x)

x = layers.GlobalAveragePooling2D()(x)
x = layers.Dropout(0.25)(x)

outputs = layers.Dense(10, activation='softmax')(x)

model_cnn = models.Model(inputs, outputs)

model_cnn.summary()

# 4. Compile
model_cnn.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

# 5. Training
history_cnn = model_cnn.fit(
    X_train, y_train,
    validation_data=(X_val, y_val),
    epochs=100,
    batch_size=128,
    callbacks=[
        EarlyStopping(
            monitor='val_loss',
            patience=15,
            restore_best_weights=True
        )
    ]
)

# 6. Prediction
y_pred_cnn = np.argmax(model_cnn.predict(X_test), axis=1)

# 7. Classification results
loss_cnn, acc_cnn = model_cnn.evaluate(X_test, y_test)

print("Test Loss:", loss_cnn)
print("Test Accuracy:", acc_cnn)
print(classification_report(y_test, y_pred_cnn))

# 8. Confusion matrix
ConfusionMatrixDisplay.from_predictions(
    y_test, y_pred_cnn, cmap='Blues'
)
plt.title('CIFAR-10 CNN Confusion Matrix')
plt.show()

# 9. Loss curve
plt.plot(history_cnn.history['loss'], label='Training Loss')
plt.plot(history_cnn.history['val_loss'], label='Validation Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()
plt.show()

print("Optimal Epoch:",
      np.argmin(history_cnn.history['val_loss']) + 1)
```

#3. Bikeshare
3-1) DNN, 1D-CNN, RF 예측 성능 비교

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow as tf

from tensorflow.keras import Sequential
from tensorflow.keras.layers import (
    Input, Dense, Conv1D, MaxPooling1D, Flatten
)
from tensorflow.keras.callbacks import EarlyStopping

from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import KFold, train_test_split
from sklearn.metrics import (
    mean_squared_error, mean_absolute_error, r2_score
)

# Hyperparameters
SEED = 123
EPOCHS = 500
BATCH = 256
LR = 0.001
PATIENCE = 20

tf.keras.utils.set_random_seed(SEED)

# Data loading
data = pd.read_csv('/content/bike.csv')

X = data[['mnth', 'hr', 'workingday', 'weathersit', 'temp']]
y = data['bikers']

# One-hot encoding
X = pd.get_dummies(
    X,
    columns=['mnth', 'hr', 'workingday', 'weathersit'],
    drop_first=True,
    dtype=float
)

feature_names = X.columns.tolist()

X = X.to_numpy(dtype='float32')
y = y.to_numpy(dtype='float32')

p = X.shape[1]

print("X shape:", X.shape)
print("y shape:", y.shape)
```

② DNN 모델 구축

```python
def build_dnn():

    model = Sequential([
        Input(shape=(p,)),
        Dense(128, activation='relu'),
        Dense(64, activation='relu'),
        Dense(1)
    ])

    model.compile(
        optimizer=tf.keras.optimizers.Adam(
            learning_rate=LR
        ),
        loss='mse'
    )

    return model

build_dnn().summary()
```

③ 1D-CNN 모델 구축

```python
def build_cnn():

    model = Sequential([
        Input(shape=(p, 1)),
        Conv1D(32, kernel_size=3, activation='relu'),
        MaxPooling1D(pool_size=2),
        Flatten(),
        Dense(64, activation='relu'),
        Dense(1)
    ])

    model.compile(
        optimizer=tf.keras.optimizers.Adam(
            learning_rate=LR
        ),
        loss='mse'
    )

    return model

build_cnn().summary()
```

④ Random Forest 모델 구축

```python
def build_rf():

    model = RandomForestRegressor(
        n_estimators=100,
        random_state=SEED,
        n_jobs=-1
    )

    return model
```

⑤ 10-fold Cross Validation

```python
kf = KFold(
    n_splits=10,
    shuffle=True,
    random_state=SEED
)

results = []

for fold, (train_idx, test_idx) in enumerate(kf.split(X), 1):

    X_train_full = X[train_idx]
    y_train_full = y[train_idx]

    X_test = X[test_idx]
    y_test = y[test_idx]

    X_train, X_val, y_train, y_val = train_test_split(
        X_train_full,
        y_train_full,
        test_size=0.2,
        random_state=SEED
    )

    # DNN
    tf.keras.utils.set_random_seed(SEED)

    dnn = build_dnn()

    dnn.fit(
        X_train, y_train,
        validation_data=(X_val, y_val),
        epochs=EPOCHS,
        batch_size=BATCH,
        callbacks=[
            EarlyStopping(
                monitor='val_loss',
                patience=PATIENCE,
                restore_best_weights=True
            )
        ],
        verbose=0
    )

    pred_dnn = dnn.predict(
        X_test, verbose=0
    ).ravel()

    # 1D-CNN
    tf.keras.utils.set_random_seed(SEED)

    cnn = build_cnn()

    cnn.fit(
        X_train.reshape(-1, p, 1),
        y_train,
        validation_data=(
            X_val.reshape(-1, p, 1), y_val
        ),
        epochs=EPOCHS,
        batch_size=BATCH,
        callbacks=[
            EarlyStopping(
                monitor='val_loss',
                patience=PATIENCE,
                restore_best_weights=True
            )
        ],
        verbose=0
    )

    pred_cnn = cnn.predict(
        X_test.reshape(-1, p, 1),
        verbose=0
    ).ravel()

    # Random Forest
    rf = build_rf()

    rf.fit(X_train, y_train)

    pred_rf = rf.predict(X_test)

    # Evaluation
    for name, pred in [
        ('DNN', pred_dnn),
        ('1D-CNN', pred_cnn),
        ('RF', pred_rf)
    ]:

        results.append({
            'Fold': fold,
            'Model': name,
            'RMSE': np.sqrt(
                mean_squared_error(y_test, pred)
            ),
            'MAE': mean_absolute_error(y_test, pred),
            'R2': r2_score(y_test, pred)
        })

    print(f'Fold {fold} completed')

results_df = pd.DataFrame(results)

print(results_df)
```

⑥ 예측 성능 비교표

```python
summary = results_df.groupby('Model')[
    ['RMSE', 'MAE', 'R2']
].agg(['mean', 'std'])

print(summary.round(4))
```

#3-2. PFI와 SHAP 변수 중요도 비교

① PFI 계산 함수

```python
def calculate_pfi(predict_fn, X_test, y_test, n_repeats=30):

    rng = np.random.default_rng(SEED)

    original_mse = mean_squared_error(
        y_test,
        predict_fn(X_test)
    )

    importance = []

    for j in range(X_test.shape[1]):

        mse_increase = []

        for _ in range(n_repeats):

            X_perm = X_test.copy()

            X_perm[:, j] = rng.permutation(
                X_perm[:, j]
            )

            pred = predict_fn(X_perm)

            perm_mse = mean_squared_error(
                y_test, pred
            )

            mse_increase.append(
                perm_mse - original_mse
            )

        importance.append(np.mean(mse_increase))

    return np.array(importance)
```

② DNN, 1D-CNN, RF의 PFI

```python
pfi_dnn = calculate_pfi(
    lambda x: dnn.predict(x, verbose=0).ravel(),
    X_test,
    y_test
)

pfi_cnn = calculate_pfi(
    lambda x: cnn.predict(
        x.reshape(-1, p, 1),
        verbose=0
    ).ravel(),
    X_test,
    y_test
)

pfi_rf = calculate_pfi(
    rf.predict,
    X_test,
    y_test
)

pfi_df = pd.DataFrame({
    'Feature': feature_names,
    'DNN': pfi_dnn,
    '1D-CNN': pfi_cnn,
    'RF': pfi_rf
})

print(pfi_df)
```

③ PFI 그래프 비교

```python
fig, axes = plt.subplots(1, 3, figsize=(18, 10))

for ax, name in zip(axes, ['DNN', '1D-CNN', 'RF']):

    temp = pfi_df.sort_values(name)

    ax.barh(temp['Feature'], temp[name])
    ax.set_title(f'{name} PFI')
    ax.set_xlabel('MSE Increase')

plt.tight_layout()
plt.show()
```

```python
# Colab에서 처음 실행하는 경우
# %pip install shap

import shap

rng = np.random.default_rng(SEED)

# Background data
background = X_train[
    rng.choice(len(X_train), 50, replace=False)
]

# SHAP evaluation data
X_explain = X_test[
    rng.choice(len(X_test), 100, replace=False)
]
```

```python
explainer_dnn = shap.Explainer(
    lambda x: dnn.predict(x, verbose=0).ravel(),
    background,
    algorithm='permutation',
    feature_names=feature_names,
    seed=SEED
)

shap_dnn = explainer_dnn(
    X_explain,
    max_evals=10 * (2 * p + 1)
)

shap.plots.bar(shap_dnn)

shap.plots.beeswarm(shap_dnn)

shap.plots.waterfall(shap_dnn[0])
```

⑥ 1D-CNN SHAP

```python
explainer_cnn = shap.Explainer(
    lambda x: cnn.predict(
        x.reshape(-1, p, 1),
        verbose=0
    ).ravel(),
    background,
    algorithm='permutation',
    feature_names=feature_names,
    seed=SEED
)

shap_cnn = explainer_cnn(
    X_explain,
    max_evals=10 * (2 * p + 1)
)

shap.plots.bar(shap_cnn)

shap.plots.beeswarm(shap_cnn)

shap.plots.waterfall(shap_cnn[0])
```

⑦ Random Forest SHAP

```python
explainer_rf = shap.TreeExplainer(
    rf,
    data=background,
    feature_names=feature_names
)
shap_rf = explainer_rf(X_explain, check_additivity=False)

shap.plots.bar(shap_rf)
shap.plots.beeswarm(shap_rf)
shap.plots.waterfall(shap_rf[0])
```

⑧ SHAP 변수 중요도 비교

```python
shap_df = pd.DataFrame({
    'Feature': feature_names,

    'DNN': np.abs(
        shap_dnn.values
    ).mean(axis=0),

    '1D-CNN': np.abs(
        shap_cnn.values
    ).mean(axis=0),

    'RF': np.abs(
        shap_rf.values
    ).mean(axis=0)
})

print(shap_df)
```

```python
fig, axes = plt.subplots(1, 3, figsize=(18, 10))

for ax, name in zip(axes, ['DNN', '1D-CNN', 'RF']):

    temp = shap_df.sort_values(name)

    ax.barh(temp['Feature'], temp[name])
    ax.set_title(f'{name} SHAP')
    ax.set_xlabel('Mean Absolute SHAP')

plt.tight_layout()
plt.show()
```

PFI: 특정 변수가 예측 성능에 얼마나 중요한지 평가합니다.
SHAP: 특정 변수가 자전거 대여 예측값을 얼마나 증가 또는 감소시키는지 평가합니다.
Bar Plot: 전체 변수의 중요도를 비교합니다.
Beeswarm Plot: 변수의 값과 예측 기여 방향을 분석합니다.
Waterfall Plot: 개별 관측치의 예측값이 만들어진 과정을 설명합니다.
